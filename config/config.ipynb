```python
# =============================================================================
# Notebook: 00_config
# Path:     /Repos/awproject/databricks_learning/config/00_config
# Purpose:  Central config — all environments in one place
# Usage:    %run /Repos/awproject/databricks_learning/config/00_config
# =============================================================================

# =============================================================================
# WIDGET — only env is needed, everything else comes from the config dict
# =============================================================================
import os

# Reads DATABRICKS_ENV set in cluster environment variables
ENV = os.environ.get("DATABRICKS_ENV", "dev")

#ENV = dbutils.widgets.get("env")

#dbutils.widgets.dropdown("env", "dev", ["dev", "qa", "prod"])

#ENV = dbutils.widgets.get("env")

# =============================================================================
# ENVIRONMENT CONFIGS
# Add or update any environment here — no other file needs to change
# =============================================================================
ENV_CONFIGS = {
    "dev": {
        "adls_account": "awprojectadls0506",
        "container":    "source",
        "base_path":    "Returns",
        "catalog":      "demo",
    },
    "qa": {
        "adls_account": "awprojectadlsqa",
        "container":    "source",
        "base_path":    "Returns",
        "catalog":      "demo_qa",
    },
    "prod": {
        "adls_account": "awprojectadlsprod",
        "container":    "source",
        "base_path":    "Returns",
        "catalog":      "demo_prod",
    },
}

# =============================================================================
# VALIDATE — fail fast if env is not recognised
# =============================================================================
if ENV not in ENV_CONFIGS:
    raise ValueError(f"Unknown env '{ENV}'. Must be one of: {list(ENV_CONFIGS.keys())}")

# =============================================================================
# RESOLVE VARIABLES FROM SELECTED ENV
# =============================================================================
cfg          = ENV_CONFIGS[ENV]
ADLS_ACCOUNT = cfg["adls_account"]
CONTAINER    = cfg["container"]
BASE_PATH    = cfg["base_path"]
CATALOG      = cfg["catalog"]

ADLS_BASE = f"abfss://{CONTAINER}@{ADLS_ACCOUNT}.dfs.core.windows.net/{BASE_PATH}"
BRONZE    = f"{CATALOG}.bronze"
SILVER    = f"{CATALOG}.silver"
GOLD      = f"{CATALOG}.gold"
# BATCH_CONTROL_TABLE is defined in 00_batch_control → demo.pipeline.batch_control

# Create all schemas if they do not exist
for schema in [BRONZE, SILVER, GOLD]:
    spark.sql(f"CREATE SCHEMA IF NOT EXISTS {schema}")

print(f"ENV      : {ENV}")
print(f"ADLS_BASE: {ADLS_BASE}")
print(f"BRONZE   : {BRONZE}")
print(f"SILVER   : {SILVER}")
print(f"GOLD     : {GOLD}")
```
