```python

from pyspark.sql import functions as F
from functools import reduce


# =============================================================================
# CELL 4 — helpers
# =============================================================================

def read_csv(filename: str):
    return (
        spark.read
        .format("csv")
        .option("header",      "true")
        .option("inferSchema", "false")
        .option("nullValue",   "")
        .option("emptyValue",  "")
        .load(f"{ADLS_BASE}/{filename}")
    )


def add_audit_cols(df, source_file: str):
    return (
        df
        .withColumn("source_file",        F.lit(source_file))
        .withColumn("ingestion_timestamp", F.current_timestamp())
        .withColumn("ingestion_date",      F.current_date())
        .withColumn("batch_id",            F.date_format(F.current_timestamp(), "yyyyMMddHHmmss"))
    )


def write_bronze(df, table_name: str):
    """All file sources are FULL load — always overwrite."""
    (
        df.write
        .format("delta")
        .mode("overwrite")
        .option("overwriteSchema", "true")
        .saveAsTable(f"{BRONZE}.{table_name}")
    )
    return df.count()


# =============================================================================
# CELL 5 — calendar
# =============================================================================
table_name  = "calendar"
source_file = "AdventureWorks_Calendar.csv"
ctrl        = get_last_load(table_name, "bronze")
load_type   = ctrl["load_type"]
status      = ctrl["status"]

if status == "SUCCESS":
    print(f"[SKIP] {table_name} — already SUCCESS")
else:
    print(f"[{load_type}] {BRONZE}.{table_name} | {source_file} | prev status: {status}")
    log_batch_start(table_name, "bronze", load_type)
    try:
        df   = add_audit_cols(read_csv(source_file), source_file)
        rows = write_bronze(df, table_name)
        log_batch_end(table_name, "bronze", None, rows, "SUCCESS")
    except Exception as e:
        log_batch_end(table_name, "bronze", None, 0, "FAILED")
        raise


# =============================================================================
# CELL 6 — customers
# =============================================================================
table_name  = "customers"
source_file = "AdventureWorks_Customers.csv"
ctrl        = get_last_load(table_name, "bronze")
load_type   = ctrl["load_type"]
status      = ctrl["status"]

if status == "SUCCESS":
    print(f"[SKIP] {table_name} — already SUCCESS")
else:
    print(f"[{load_type}] {BRONZE}.{table_name} | {source_file} | prev status: {status}")
    log_batch_start(table_name, "bronze", load_type)
    try:
        df   = add_audit_cols(read_csv(source_file), source_file)
        rows = write_bronze(df, table_name)
        log_batch_end(table_name, "bronze", None, rows, "SUCCESS")
    except Exception as e:
        log_batch_end(table_name, "bronze", None, 0, "FAILED")
        raise


# =============================================================================
# CELL 7 — product_categories
# =============================================================================
table_name  = "product_categories"
source_file = "AdventureWorks_Product_Categories.csv"
ctrl        = get_last_load(table_name, "bronze")
load_type   = ctrl["load_type"]
status      = ctrl["status"]

if status == "SUCCESS":
    print(f"[SKIP] {table_name} — already SUCCESS")
else:
    print(f"[{load_type}] {BRONZE}.{table_name} | {source_file} | prev status: {status}")
    log_batch_start(table_name, "bronze", load_type)
    try:
        df   = add_audit_cols(read_csv(source_file), source_file)
        rows = write_bronze(df, table_name)
        log_batch_end(table_name, "bronze", None, rows, "SUCCESS")
    except Exception as e:
        log_batch_end(table_name, "bronze", None, 0, "FAILED")
        raise


# =============================================================================
# CELL 8 — product_subcategories
# =============================================================================
table_name  = "product_subcategories"
source_file = "AdventureWorks_Product_Subcategories.csv"
ctrl        = get_last_load(table_name, "bronze")
load_type   = ctrl["load_type"]
status      = ctrl["status"]

if status == "SUCCESS":
    print(f"[SKIP] {table_name} — already SUCCESS")
else:
    print(f"[{load_type}] {BRONZE}.{table_name} | {source_file} | prev status: {status}")
    log_batch_start(table_name, "bronze", load_type)
    try:
        df   = add_audit_cols(read_csv(source_file), source_file)
        rows = write_bronze(df, table_name)
        log_batch_end(table_name, "bronze", None, rows, "SUCCESS")
    except Exception as e:
        log_batch_end(table_name, "bronze", None, 0, "FAILED")
        raise


# =============================================================================
# CELL 9 — products
# =============================================================================
table_name  = "products"
source_file = "AdventureWorks_Products.csv"
ctrl        = get_last_load(table_name, "bronze")
load_type   = ctrl["load_type"]
status      = ctrl["status"]

if status == "SUCCESS":
    print(f"[SKIP] {table_name} — already SUCCESS")
else:
    print(f"[{load_type}] {BRONZE}.{table_name} | {source_file} | prev status: {status}")
    log_batch_start(table_name, "bronze", load_type)
    try:
        df   = add_audit_cols(read_csv(source_file), source_file)
        rows = write_bronze(df, table_name)
        log_batch_end(table_name, "bronze", None, rows, "SUCCESS")
    except Exception as e:
        log_batch_end(table_name, "bronze", None, 0, "FAILED")
        raise


# =============================================================================
# CELL 10 — territories
# =============================================================================
table_name  = "territories"
source_file = "AdventureWorks_Territories.csv"
ctrl        = get_last_load(table_name, "bronze")
load_type   = ctrl["load_type"]
status      = ctrl["status"]

if status == "SUCCESS":
    print(f"[SKIP] {table_name} — already SUCCESS")
else:
    print(f"[{load_type}] {BRONZE}.{table_name} | {source_file} | prev status: {status}")
    log_batch_start(table_name, "bronze", load_type)
    try:
        df   = add_audit_cols(read_csv(source_file), source_file)
        rows = write_bronze(df, table_name)
        log_batch_end(table_name, "bronze", None, rows, "SUCCESS")
    except Exception as e:
        log_batch_end(table_name, "bronze", None, 0, "FAILED")
        raise


# =============================================================================
# CELL 11 — sales (union 2015 + 2016 + 2017)
# =============================================================================
table_name   = "sales"
source_files = [
    "AdventureWorks_Sales_2015.csv",
    "AdventureWorks_Sales_2016.csv",
    "AdventureWorks_Sales_2017.csv",
]
ctrl         = get_last_load(table_name, "bronze")
load_type    = ctrl["load_type"]
status       = ctrl["status"]

if status == "SUCCESS":
    print(f"[SKIP] {table_name} — already SUCCESS")
else:
    print(f"[{load_type}] {BRONZE}.{table_name} | {source_files} | prev status: {status}")
    log_batch_start(table_name, "bronze", load_type)
    try:
        df = reduce(
            lambda a, b: a.unionByName(b),
            [read_csv(f) for f in source_files]
        )
        df   = add_audit_cols(df, ",".join(source_files))
        rows = write_bronze(df, table_name)
        log_batch_end(table_name, "bronze", None, rows, "SUCCESS")
    except Exception as e:
        log_batch_end(table_name, "bronze", None, 0, "FAILED")
        raise


# =============================================================================
# CELL 12 — summary
# =============================================================================
show_control_table("bronze")
```
